# CSP Growth: My Journey So Far

## Personal Growth

I started this class at ground zero. I knew a little bit about coding from last year, but I didn't really know how to truly work like a developer on real things other than just games like last year. Ground Zero taught me that growth isn't about being perfect, it's about showing up consistently. Every week I read the sprints, set small goals, checked in on myself and my teammates, and looked at what I could do better each week. The biggest thing I learned is that being honest about my weak spots made me improve way faster than pretending I had it all figured out. Peer feedback was also very helpful to my growth in this class. Being able to have my teammate rate how I have been doing and working on it was amazing.

## MCQ

The MCQ really showed me where my knowledge had some gaps in it. It showed me that I still had more room to grow and become better. I got some questions wrong that I thought I knew, and that was a wake-up call for me. But instead of getting mad, I went back and looked at the questions I got wrong and I tried to figure out *why* I missed them. Now I read questions slower and try to actually understand them, I think about what the question is really asking, and I connect each concept back to what we actually built in class.

## Team Growth Blog

Working with a team taught me stuff that coding alone never could. Communication is everything. When me and my teammate checked in with each other regularly, split up tasks clearly, and held each other accountable for our mistakes, we started doing a lot better in the class. When we didn't do that, things got messy real quick. I learned to ask for help early, give feedback kindly, and celebrate small wins with the team. The code matters, but the people building it matter more.

## What's Next

Ground Zero made me more self-aware, the MCQ made me more humble, and the team helped me become more collaborative. Next, I'm focusing on staying consistent, owning my mistakes faster, and helping my teammates level up too. We're not at ground zero anymore.
