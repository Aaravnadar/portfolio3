### Iteration

**CPT Requirement:** the algorithm must include a loop. Log expenses until a -1 sentinel appears.

```python
%%js

// CODE_RUNNER: Add expenses to the ledger until a -1 sentinel value appears.

const ledger = [];
const nextExpenses = [8700, 11200, 9800, -1];
let i = 0;

while (nextExpenses[i] !== -1) {
  ledger.push(nextExpenses[i]);
  i++;
}

console.log("Expense ledger: " + ledger);
```

```python
%%python

# CODE_RUNNER: Add expenses to the ledger until a -1 sentinel value appears.

ledger = []
next_expenses = [8700, 11200, 9800, -1]
i = 0

while next_expenses[i] != -1:
    ledger.append(next_expenses[i])
    i += 1

print("Expense ledger: " + str(ledger))
```

### Algorithm

**CPT Requirement:** combine sequence, selection, and iteration. Average monthly revenue, then rate it.

```python
%%js

// CODE_RUNNER: Average monthly revenue, then assign a rating.

function averageRevenue(months) {
  let total = 0;
  for (const m of months) {
    total += m;
  }
  return total / months.length;
}

const monthlyRevenue = [92000, 145000, 118000, 133000];
const avg = averageRevenue(monthlyRevenue);

let rating = "Needs Review";
if (avg >= 130000) {
  rating = "Excellent";
} else if (avg >= 100000) {
  rating = "Strong";
}

console.log("Average Revenue: $" + avg);
console.log("Rating: " + rating);
```

```python
%%python

# CODE_RUNNER: Average monthly revenue, then assign a rating.

def average_revenue(months):
    total = 0
    for m in months:
        total += m
    return total / len(months)

monthly_revenue = [92000, 145000, 118000, 133000]
avg = average_revenue(monthly_revenue)

rating = "Needs Review"
if avg >= 130000:
    rating = "Excellent"
elif avg >= 100000:
    rating = "Strong"

print("Average Revenue: $" + str(avg))
print("Rating: " + rating)
```

### Search Algorithm

**Exam Topic:** linear search. Look up a rep's sales by name.

```python
%%js

// CODE_RUNNER: Search the roster for a rep's sales using linear search.

function findSales(names, sales, target) {
  for (let i = 0; i < names.length; i++) {
    if (names[i] === target) {
      return sales[i];
    }
  }
  return -1;
}

const names = ["Dana Reyes", "Luis Ortega", "Priya Shah", "Tom Becker"];
const sales = [154000, 98000, 112000, 137500];

const searchName = "Priya Shah";
const result = findSales(names, sales, searchName);

if (result !== -1) {
  console.log(searchName + "'s sales: $" + result);
} else {
  console.log(searchName + " is not on the roster.");
}
```

```python
%%python

# CODE_RUNNER: Search the roster for a rep's sales using linear search.

def find_sales(names, sales, target):
    for i in range(len(names)):
        if names[i] == target:
            return sales[i]
    return -1

names = ["Dana Reyes", "Luis Ortega", "Priya Shah", "Tom Becker"]
sales = [154000, 98000, 112000, 137500]

search_name = "Priya Shah"
result = find_sales(names, sales, search_name)

if result != -1:
    print(search_name + "'s sales: $" + str(result))
else:
    print(search_name + " is not on the roster.")
```

### Boolean Logic

**Exam Concept:** combine conditions with AND, OR, and NOT. Decide whether a deal gets approved.

```python
%%js

// CODE_RUNNER: Decide if a deal is approved using AND / OR / NOT.

const creditScore = 720;
const yearsActive = 3;
const hasCollateral = false;

const dealApproved = (creditScore >= 700 && yearsActive > 0) || hasCollateral;
const isNewBusiness = !(yearsActive > 1);

console.log("Deal approved: " + dealApproved);
console.log("New business: " + isNewBusiness);
```

```python
%%python

# CODE_RUNNER: Decide if a deal is approved using AND / OR / NOT.

credit_score = 720
years_active = 3
has_collateral = False

deal_approved = (credit_score >= 700 and years_active > 0) or has_collateral
is_new_business = not (years_active > 1)

print("Deal approved: " + str(deal_approved))
print("New business: " + str(is_new_business))
```
