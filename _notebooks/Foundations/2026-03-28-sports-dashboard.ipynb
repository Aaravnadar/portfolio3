### Procedure

**CPT Requirement:** one student-developed procedure with a parameter and a return value. Turn points into a player tier.

```python

# CODE_RUNNER: Write a procedure that returns a player tier based on points.

def get_player_tier(points):
    if points >= 300:
        return "All-Star"
    elif points >= 200:
        return "Starter"
    else:
        return "Bench"

player_quarter_points = 275
print("Tier: " + get_player_tier(player_quarter_points))
```

### Sequence

**Exam Concept:** statements run in order. Build a final game score step by step.

```python

# CODE_RUNNER: Calculate a final game score step-by-step.

base_points = 82
bonus_points = 15
foul_penalty = 6
overtime_multiplier = 1.1

pre_bonus = base_points + bonus_points - foul_penalty
final_score = pre_bonus * overtime_multiplier

print("Final Score: " + str(final_score))
```

### Selection

**CPT Requirement:** the algorithm must include IF / ELSE decisions. Rate a player on season points.

```python

# CODE_RUNNER: Assign a performance rating based on season points.

season_points_total = 1375
rating = "Needs Review"

if season_points_total >= 1500:
    rating = "Excellent"
elif season_points_total >= 1200:
    rating = "Strong"
elif season_points_total >= 900:
    rating = "Fair"

print("Performance rating: " + rating)
```

### Iteration

**CPT Requirement:** the algorithm must include a loop. Log game scores until a -1 sentinel appears.

```python

# CODE_RUNNER: Add game scores to the log until a -1 sentinel value appears.

game_log = []
next_scores = [24, 31, 28, -1]
i = 0

while next_scores[i] != -1:
    game_log.append(next_scores[i])
    i += 1

print("Game log: " + str(game_log))
```

### Algorithm

**CPT Requirement:** combine sequence, selection, and iteration. Average points per game, then rate it.

```python

# CODE_RUNNER: Average points per game, then assign a rating.

def average_points(games):
    total = 0
    for g in games:
        total += g
    return total / len(games)

game_points = [22, 34, 27, 30]
avg = average_points(game_points)

rating = "Needs Review"
if avg >= 30:
    rating = "Excellent"
elif avg >= 25:
    rating = "Strong"

print("Average Points: " + str(avg))
print("Rating: " + rating)
```

### Search Algorithm

**Exam Topic:** linear search. Look up a player's points by name.

```python

# CODE_RUNNER: Search the roster for a player's points using linear search.

def find_points(names, points, target):
    for i in range(len(names)):
        if names[i] == target:
            return points[i]
    return -1

names = ["Jordan Lee", "Sam Diaz", "Taylor Reed", "Alex Kim"]
points = [1540, 980, 1120, 1375]

search_name = "Taylor Reed"
result = find_points(names, points, search_name)

if result != -1:
    print(search_name + "'s points: " + str(result))
else:
    print(search_name + " is not on the roster.")
```

### Boolean Logic

**Exam Concept:** combine conditions with AND, OR, and NOT. Decide whether a player makes the All-Star team.

```python

# CODE_RUNNER: Decide if a player is an All-Star using AND / OR / NOT.

points_per_game = 28
games_played = 15
coach_recommendation = False

all_star_selected = (points_per_game >= 25 and games_played > 10) or coach_recommendation
is_rookie = not (games_played > 5)

print("All-Star selected: " + str(all_star_selected))
print("Rookie: " + str(is_rookie))
```
