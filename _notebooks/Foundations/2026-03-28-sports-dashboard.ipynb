## 2. Input

**Basketball example:** Check-in questions when a new player joins the roster.

Input lets a person hand data to your program. The `ask()` helper stands in for `input()` and `prompt()` so the cell never freezes.

**Why it matters for the exam:** Input is required for the Create Performance Task. It shows your program reacts to different data.


### Python

```python
# CODE_RUNNER: Ask for the player name and starting points.

answers = ["Jordan Lee", "0"]
def ask(question):
    reply = answers.pop(0)
    print(question + " " + reply)
    return reply

player_name_input = ask("Enter player name:")
starting_points = ask("Enter starting points:")

print("Player " + player_name_input + " starts with " + starting_points + " points.")
```

### JavaScript

```python
%%js
// CODE_RUNNER: Ask for the player name and starting points.

let answers = ["Jordan Lee", "0"];
function ask(question) {
  let reply = answers.shift();
  console.log(question + " " + reply);
  return reply;
}

let playerNameInput = ask("Enter player name:");
let startingPoints = ask("Enter starting points:");

console.log("Player " + playerNameInput + " starts with " + startingPoints + " points.");
```

### Pseudocode

```python
%%pseudocode
// CODE_RUNNER: Ask for the player name and starting points.

playerNameInput <-- INPUT("Enter player name:")
startingPoints <-- INPUT("Enter starting points:")

DISPLAY("Player " + playerNameInput + " starts with " + startingPoints + " points.")
```

## 3. List and Traversal

**Basketball example:** Adding up this season's points, quarter by quarter.

A list holds many values under one name. A traversal walks through every item once. Here each number is one quarter's worth of points.

**Why it matters for the exam:** Using a list to manage data earns points for handling complexity, and exam questions love a loop that builds a running total.


### Python

```python
# CODE_RUNNER: Store this season's points per quarter and total them.

quarterly_points = [220, 310, 275, 340]
total = 0

for amount in quarterly_points:
    total += amount

print("Total points this season: " + str(total))
```

### JavaScript

```python
%%js
// CODE_RUNNER: Store this season's points per quarter and total them.

const quarterlyPoints = [220, 310, 275, 340];
let total = 0;

for (const amount of quarterlyPoints) {
  total += amount;
}

console.log("Total points this season: " + total);
```

### Pseudocode

```python
%%pseudocode
// CODE_RUNNER: Store this season's points per quarter and total them.

quarterlyPoints <-- [220, 310, 275, 340]
total <-- 0

FOR EACH amount IN quarterlyPoints {
    total <-- total + amount
}

DISPLAY("Total points this season: " + total)
```

## 4. Procedure with a Parameter and a Return Value

**Basketball example:** Turning a quarter's points into a player tier.

A procedure is a named block you can call over and over. It takes a parameter and hands back an answer with a return.

**Why it matters for the exam:** You must write at least one procedure that takes a parameter and gets called at least once. This is the abstraction point on the scoring guide.


### Python

```python
# CODE_RUNNER: Write a procedure that returns a player tier based on points.

def get_player_tier(points):
    if points >= 300:
        return "All-Star"
    elif points >= 200:
        return "Starter"
    else:
        return "Bench"

player_quarter_points = 275
print("Tier: " + get_player_tier(player_quarter_points))
```

### JavaScript

```python
%%js
// CODE_RUNNER: Write a procedure that returns a player tier based on points.

function getPlayerTier(points) {
  if (points >= 300) return "All-Star";
  if (points >= 200) return "Starter";
  return "Bench";
}

console.log("Tier: " + getPlayerTier(275));
```

### Pseudocode

```python
%%pseudocode
// CODE_RUNNER: Write a procedure that returns a player tier based on points.

PROCEDURE getPlayerTier(points) {
    IF (points >= 300) {
        RETURN "All-Star"
    } ELSE {
        IF (points >= 200) {
            RETURN "Starter"
        } ELSE {
            RETURN "Bench"
        }
    }
}

DISPLAY(getPlayerTier(275))
```

## 5. Sequence

**Basketball example:** Building a final game score step by step.

Sequence means the computer runs your lines in order, top to bottom. Nothing jumps around or repeats.

**Why it matters for the exam:** Every program uses sequence. You will trace line by line and predict a variable's value at a certain point.


### Python

```python
# CODE_RUNNER: Calculate a final game score step-by-step.

base_points = 82
bonus_points = 15
foul_penalty = 6
overtime_multiplier = 1.1

pre_bonus = base_points + bonus_points - foul_penalty
final_score = pre_bonus * overtime_multiplier

print("Final Score: " + str(final_score))
```

### JavaScript

```python
%%js
// CODE_RUNNER: Calculate a final game score step-by-step.

const basePoints = 82;
const bonusPoints = 15;
const foulPenalty = 6;
const overtimeMultiplier = 1.1;

const preBonus = basePoints + bonusPoints - foulPenalty;
const finalScore = preBonus * overtimeMultiplier;

console.log("Final Score: " + finalScore);
```

### Pseudocode

```python
%%pseudocode
// CODE_RUNNER: Calculate a final game score step-by-step.

basePoints <-- 82
bonusPoints <-- 15
foulPenalty <-- 6
overtimeMultiplier <-- 1.1

preBonus <-- basePoints + bonusPoints - foulPenalty
finalScore <-- preBonus * overtimeMultiplier

DISPLAY("Final Score: " + finalScore)
```

## 6. Selection

**Basketball example:** Rating a player based on season points.

Selection means the program picks one path based on a true or false test. Only one branch runs.

**Why it matters for the exam:** Selection is required in your algorithm. Trace the conditions in order and stop at the first one that is true.


### Python

```python
# CODE_RUNNER: Assign a performance rating based on season points.

season_points_total = 1375
rating = "Needs Review"

if season_points_total >= 1500:
    rating = "Excellent"
elif season_points_total >= 1200:
    rating = "Strong"
elif season_points_total >= 900:
    rating = "Fair"

print("Performance rating: " + rating)
```

### JavaScript

```python
%%js
// CODE_RUNNER: Assign a performance rating based on season points.

const seasonPointsTotal = 1375;
let rating = "Needs Review";

if (seasonPointsTotal >= 1500) rating = "Excellent";
else if (seasonPointsTotal >= 1200) rating = "Strong";
else if (seasonPointsTotal >= 900) rating = "Fair";

console.log("Performance rating: " + rating);
```

### Pseudocode

```python
%%pseudocode
// CODE_RUNNER: Assign a performance rating based on season points.

seasonPointsTotal <-- 1375
rating <-- "Needs Review"

IF (seasonPointsTotal >= 1500) {
    rating <-- "Excellent"
} ELSE {
    IF (seasonPointsTotal >= 1200) {
        rating <-- "Strong"
    } ELSE {
        IF (seasonPointsTotal >= 900) {
            rating <-- "Fair"
        }
    }
}

DISPLAY("Performance rating: " + rating)
```

## 7. Iteration with a Sentinel Loop

**Basketball example:** Logging game scores until a -1 sentinel appears.

Iteration means repeating a block of code. A sentinel loop keeps going until it sees a stop value, which here is -1. Nobody knows ahead of time how many games are left to log.

**Why it matters for the exam:** Iteration is required in your algorithm. Exam questions ask how many times the loop body runs, so count the passes and watch the sentinel.


### Python

```python
# CODE_RUNNER: Add game scores to the log until a -1 sentinel value appears.

game_log = []
next_scores = [24, 31, 28, -1]
i = 0

while next_scores[i] != -1:
    game_log.append(next_scores[i])
    i += 1

print("Game log: " + str(game_log))
```

### JavaScript

```python
%%js
// CODE_RUNNER: Add game scores to the log until a -1 sentinel value appears.

const gameLog = [];
const nextScores = [24, 31, 28, -1];
let i = 0;

while (nextScores[i] !== -1) {
  gameLog.push(nextScores[i]);
  i++;
}

console.log("Game log: " + gameLog);
```

### Pseudocode

```python
%%pseudocode
// CODE_RUNNER: Add game scores to the log until a -1 sentinel value appears.

gameLog <-- []
nextScores <-- [24, 31, 28, -1]
i <-- 1

REPEAT UNTIL (nextScores[i] = -1) {
    APPEND(gameLog, nextScores[i])
    i <-- i + 1
}

DISPLAY(gameLog)
```

## 8. Algorithm, All Three Pieces Together

**Basketball example:** Averaging points per game, then assigning a rating.

An algorithm puts sequence, selection, and iteration together. This one loops through the games, totals the points, then judges the result.

**Why it matters for the exam:** This is the algorithm implementation point. One loop alone is not enough and one if statement alone is not enough.


### Python

```python
# CODE_RUNNER: Average points per game, then assign a rating.

def average_points(games):
    total = 0
    for g in games:
        total += g
    return total / len(games)

game_points = [22, 34, 27, 30]
avg = average_points(game_points)

rating = "Needs Review"
if avg >= 30:
    rating = "Excellent"
elif avg >= 25:
    rating = "Strong"

print("Average Points: " + str(avg))
print("Rating: " + rating)
```

### JavaScript

```python
%%js
// CODE_RUNNER: Average points per game, then assign a rating.

function averagePoints(games) {
  let total = 0;
  for (const game of games) total += game;
  return total / games.length;
}

const average = averagePoints([22, 34, 27, 30]);
const rating = average >= 30 ? "Excellent" : average >= 25 ? "Strong" : "Needs Review";
console.log("Average Points: " + average);
console.log("Rating: " + rating);
```

### Pseudocode

```python
%%pseudocode
// CODE_RUNNER: Average points per game, then assign a rating.

PROCEDURE averagePoints(games) {
    total <-- 0
    FOR EACH game IN games {
        total <-- total + game
    }
    RETURN total / LENGTH(games)
}

average <-- averagePoints([22, 34, 27, 30])
rating <-- "Needs Review"
IF (average >= 30) {
    rating <-- "Excellent"
} ELSE {
    IF (average >= 25) {
        rating <-- "Strong"
    }
}
DISPLAY("Average Points: " + average)
DISPLAY("Rating: " + rating)
```

## 9. List Operations

**Basketball example:** Managing the starting lineup.

These are the four list operations the exam expects: add to the end, insert at a position, remove at a position, and get the length.

**Why it matters for the exam:** Removing an item slides every later item down by one, and that shift is a favorite trap.


### Python

```python
# CODE_RUNNER: Add, insert, and remove players from the lineup.

lineup = ["Alex Kim", "Sam Diaz"]
print("Initial: " + str(lineup))

lineup.append("Jordan Lee")
print("After append: " + str(lineup))

lineup.insert(1, "Taylor Reed")
print("After insert at 1: " + str(lineup))

lineup.pop(2)
print("After remove at 2: " + str(lineup))

print("Lineup size: " + str(len(lineup)))
```

### JavaScript

```python
%%js
// CODE_RUNNER: Add, insert, and remove players from the lineup.

const lineup = ["Alex Kim", "Sam Diaz"];
console.log("Initial: " + lineup);

lineup.push("Jordan Lee");
console.log("After append: " + lineup);

lineup.splice(1, 0, "Taylor Reed");
console.log("After insert at 1: " + lineup);

lineup.splice(2, 1);
console.log("After remove at 2: " + lineup);

console.log("Lineup size: " + lineup.length);
```

### Pseudocode

```python
%%pseudocode
// CODE_RUNNER: Add, insert, and remove players from the lineup.

lineup <-- ["Alex Kim", "Sam Diaz"]
lineup <-- APPEND(lineup, "Jordan Lee")
lineup <-- INSERT(lineup, 2, "Taylor Reed")
lineup <-- REMOVE(lineup, 3)
DISPLAY(lineup)
DISPLAY(LENGTH(lineup))
```

## 10. Linear Search

**Basketball example:** Looking up a player's points by name.

A linear search checks each item in order until it finds a match. If it runs off the end it returns -1, the normal way for code to say not found.

**Why it matters for the exam:** Know what a linear search returns when the item is missing, and know that the worst case checks every item.


### Python

```python
# CODE_RUNNER: Search the roster for a player's points using linear search.

def find_points(names, points, target):
    for i in range(len(names)):
        if names[i] == target:
            return points[i]
    return -1

names = ["Jordan Lee", "Sam Diaz", "Taylor Reed", "Alex Kim"]
points = [1540, 980, 1120, 1375]

search_name = "Taylor Reed"
result = find_points(names, points, search_name)

if result != -1:
    print(search_name + "'s points: " + str(result))
else:
    print(search_name + " is not on the roster.")
```

### JavaScript

```python
%%js
// CODE_RUNNER: Search the roster for a player's points using linear search.

function findPoints(names, points, target) {
  for (let i = 0; i < names.length; i++) {
    if (names[i] === target) return points[i];
  }
  return -1;
}

const names = ["Jordan Lee", "Sam Diaz", "Taylor Reed", "Alex Kim"];
const points = [1540, 980, 1120, 1375];
const searchName = "Taylor Reed";
const result = findPoints(names, points, searchName);
console.log(result !== -1 ? searchName + "'s points: " + result : searchName + " is not on the roster.");
```

### Pseudocode

```python
%%pseudocode
// CODE_RUNNER: Search the roster for a player's points using linear search.

PROCEDURE findPoints(names, points, target) {
    i <-- 1
    FOR EACH name IN names {
        IF (name = target) {
            RETURN points[i]
        }
        i <-- i + 1
    }
    RETURN -1
}

names <-- ["Jordan Lee", "Sam Diaz", "Taylor Reed", "Alex Kim"]
points <-- [1540, 980, 1120, 1375]
result <-- findPoints(names, points, "Taylor Reed")
DISPLAY(result)
```

## 11. Boolean Logic

**Basketball example:** Deciding whether a player makes the All-Star team.

Boolean logic combines conditions with AND, OR, and NOT to produce a single true or false answer.

**Why it matters for the exam:** Exam questions build truth tables from these operators, so know how AND, OR, and NOT combine and short-circuit.


### Python

```python
# CODE_RUNNER: Decide if a player is an All-Star using AND / OR / NOT.

points_per_game = 28
games_played = 15
coach_recommendation = False

all_star_selected = (points_per_game >= 25 and games_played > 10) or coach_recommendation
is_rookie = not (games_played > 5)

print("All-Star selected: " + str(all_star_selected))
print("Rookie: " + str(is_rookie))
```

### JavaScript

```python
%%js
// CODE_RUNNER: Decide if a player is an All-Star using AND / OR / NOT.

const pointsPerGame = 28;
const gamesPlayed = 15;
const coachRecommendation = false;
const allStarSelected = (pointsPerGame >= 25 && gamesPlayed > 10) || coachRecommendation;
const isRookie = !(gamesPlayed > 5);
console.log("All-Star selected: " + allStarSelected);
console.log("Rookie: " + isRookie);
```

### Pseudocode

```python
%%pseudocode
// CODE_RUNNER: Decide if a player is an All-Star using AND / OR / NOT.

pointsPerGame <-- 28
gamesPlayed <-- 15
coachRecommendation <-- FALSE
allStarSelected <-- (pointsPerGame >= 25 AND gamesPlayed > 10) OR coachRecommendation
isRookie <-- NOT (gamesPlayed > 5)
DISPLAY("All-Star selected: " + allStarSelected)
DISPLAY("Rookie: " + isRookie)
```

## Side by Side Cheat Sheet

| Idea | Pseudocode | Python | JavaScript |
|------|-----------|--------|------------|
| Assignment | `x <-- 5` | `x = 5` | `let x = 5;` |
| Output | `DISPLAY(x)` | `print(x)` | `console.log(x)` |
| Input | `INPUT("msg")` | `input("msg")` | `prompt("msg")` |
| First list position | `1` | `0` | `0` |
| Procedure | `PROCEDURE f(a)` | `def f(a):` | `function f(a) {}` |
| Block markers | `{ }` | indentation | `{ }` |
| Equality test | `=` | `==` | `===` |
| Not equal | `NOT (a = b)` | `a != b` | `a !== b` |
| Add to the end | `APPEND(l, v)` | `l.append(v)` | `l.push(v)` |
| Insert at position i | `INSERT(l, i, v)` | `l.insert(i-1, v)` | `l.splice(i-1, 0, v)` |
| Remove at position i | `REMOVE(l, i)` | `l.pop(i-1)` | `l.splice(i-1, 1)` |
| Length | `LENGTH(l)` | `len(l)` | `l.length` |
| Loop over a list | `FOR EACH x IN l` | `for x in l:` | `for (let x of l)` |
| Loop until | `REPEAT UNTIL (c)` | `while not c:` | `while (!c) {}` |
| Text from a number | automatic | needs `str(x)` | automatic |

**The two things that trip people up most:**

1. Python will not glue a number onto a string with `+`. Wrap it in `str()` first. JavaScript converts it
   for you without asking.
2. Python has no curly braces. The indentation is the block. Line up your spaces or the code breaks.
