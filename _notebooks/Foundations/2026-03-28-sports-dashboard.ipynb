## 3. List and Traversal

## 7. Iteration with a Sentinel Loop

Basketball example: Recording game scores until a -1 sentinel appears.

Iteration repeats a block of code. A sentinel loop keeps going until it sees a special stop value. Here, `-1` means stop.

The loop does not need to know ahead of time how many game scores will be entered.

### JavaScript

**Code Runner Challenge:** Add one more game score before the -1 and run it again.


```python
%%js

// CODE_RUNNER: Add game scores to the log until a -1 sentinel appears.
const gameLog = [];
const nextScores = [24, 31, 28, -1];
let i = 0;

while (nextScores[i] !== -1) {
  gameLog.push(nextScores[i]);
  i++;
}

console.log("Game log: " + gameLog);
```

### Python

```python
# CODE_RUNNER: Add game scores to the log until a -1 sentinel appears.
game_log = []
next_scores = [24, 31, 28, -1]
i = 0

while next_scores[i] != -1:
    game_log.append(next_scores[i])
    i = i + 1

print("Game log: " + str(game_log))
```

## 8. Algorithm, All Three Pieces

Basketball example: Finding average points per game and rating the result.

An algorithm can combine **sequence, selection, and iteration** to solve a problem.

This example uses a loop to total game scores, calculates an average, and then uses selection to assign a performance rating.

### JavaScript

**Code Runner Challenge:** Change the game scores and predict whether the rating changes.


```python
%%js

// CODE_RUNNER: Average points per game, then assign a rating.
function averagePoints(games) {
  let total = 0;
  for (const game of games) {
    total = total + game;
  }
  return total / games.length;
}

const gamePoints = [22, 34, 27, 30];
const average = averagePoints(gamePoints);

let rating;
if (average >= 30) {
  rating = "Excellent";
} else if (average >= 25) {
  rating = "Strong";
} else {
  rating = "Needs Review";
}

console.log("Average Points: " + average);
console.log("Rating: " + rating);
```

### Python

```python
# CODE_RUNNER: Average points per game, then assign a rating.
def average_points(games):
    total = 0
    for game in games:
        total = total + game
    return total / len(games)

game_points = [22, 34, 27, 30]
average = average_points(game_points)

if average >= 30:
    rating = "Excellent"
elif average >= 25:
    rating = "Strong"
else:
    rating = "Needs Review"

print("Average Points: " + str(average))
print("Rating: " + rating)
```

## 9. Linear Search

Basketball example: Looking up a player's points by searching the roster.

A linear search checks items one at a time until it finds a match. If the target is not found, the procedure returns `-1`.

The search can use one list to find a position and then use that position to retrieve related information.

### JavaScript

**Code Runner Challenge:** Search for a player who is not on the roster and confirm that the result is -1.


```python
%%js

// CODE_RUNNER: Search the roster for a player's points using linear search.
function findPoints(names, points, target) {
  for (let i = 0; i < names.length; i++) {
    if (names[i] === target) {
      return points[i];
    }
  }
  return -1;
}

const names = ["Jordan Lee", "Sam Diaz", "Taylor Reed", "Alex Kim"];
const points = [1540, 980, 1120, 1375];

const searchName = "Taylor Reed";
const result = findPoints(names, points, searchName);

if (result !== -1) {
  console.log(searchName + "'s points: " + result);
} else {
  console.log(searchName + " is not on the roster.");
}
```

### Python

```python
# CODE_RUNNER: Search the roster for a player's points using linear search.
def find_points(names, points, target):
    for i in range(len(names)):
        if names[i] == target:
            return points[i]
    return -1

names = ["Jordan Lee", "Sam Diaz", "Taylor Reed", "Alex Kim"]
points = [1540, 980, 1120, 1375]

search_name = "Taylor Reed"
result = find_points(names, points, search_name)

if result != -1:
    print(search_name + "'s points: " + str(result))
else:
    print(search_name + " is not on the roster.")
```

## 11. Parallel Lists, Find the Position

Basketball example: Matching a jersey number to a player's name.

Two lists are parallel when the item at one position in the first list belongs with the item at the same position in the second list.

This procedure searches the jersey-number list, returns a position, and then uses that position to find the player's name.

### JavaScript

**Code Runner Challenge:** Look up jersey number 12 instead and confirm the player name that comes back.


```python
%%js

// CODE_RUNNER: Find the position of a jersey number, then use it to find the player.
function findJersey(jerseys, target) {
  let index = 1;
  for (const jersey of jerseys) {
    if (jersey === target) {
      return index;
    }
    index++;
  }
  return -1;
}

const jerseys = ["3", "12", "23", "30"];
const players = ["Jordan Lee", "Sam Diaz", "Taylor Reed", "Alex Kim"];

const lookingFor = "23";
const position = findJersey(jerseys, lookingFor);

if (position > 0) {
  const playerName = players[position - 1];
  console.log("Jersey found at position: " + position);
  console.log("Player: " + playerName);
} else {
  console.log("That jersey is not on the roster.");
}
```

### Python

```python
# CODE_RUNNER: Find the position of a jersey number, then use it to find the player.
def find_jersey(jerseys, target):
    index = 1
    for jersey in jerseys:
        if jersey == target:
            return index
        index = index + 1
    return -1

jerseys = ["3", "12", "23", "30"]
players = ["Jordan Lee", "Sam Diaz", "Taylor Reed", "Alex Kim"]

looking_for = "23"
position = find_jersey(jerseys, looking_for)

if position > 0:
    player_name = players[position - 1]
    print("Jersey found at position: " + str(position))
    print("Player: " + player_name)
else:
    print("That jersey is not on the roster.")
```

## 12. Parallel Lists, Return the Answer Directly

Basketball example: Turning a player's name into their season points.

This is similar to topic 11, but the procedure does the position math inside itself and returns the answer directly.

That means the calling code does not need to know where the player is stored in the list. This is a simple example of abstraction.

### JavaScript

**Code Runner Challenge:** Add another player and search for their points without exposing the list position.


```python
%%js

// CODE_RUNNER: Return a player's season points directly from parallel lists.
function findPointsDirect(players, points, target) {
  let index = 1;
  for (const player of players) {
    if (player === target) {
      return points[index - 1];
    }
    index++;
  }
  return -1;
}

const players = ["Jordan Lee", "Sam Diaz", "Taylor Reed", "Alex Kim"];
const points = [1540, 980, 1120, 1375];

const requested = "Taylor Reed";
const result = findPointsDirect(players, points, requested);

if (result !== -1) {
  console.log(requested + " has " + result + " season points.");
} else {
  console.log("Player not found.");
}
```

### Python

```python
# CODE_RUNNER: Return a player's season points directly from parallel lists.
def find_points_direct(players, points, target):
    index = 1
    for player in players:
        if player == target:
            return points[index - 1]
        index = index + 1
    return -1

players = ["Jordan Lee", "Sam Diaz", "Taylor Reed", "Alex Kim"]
points = [1540, 980, 1120, 1375]

requested = "Taylor Reed"
result = find_points_direct(players, points, requested)

if result != -1:
    print(requested + " has " + str(result) + " season points.")
else:
    print("Player not found.")
```
