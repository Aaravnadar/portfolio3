### Iteration

**CPT Requirement:** the algorithm must include a loop. Log game scores until a -1 sentinel appears.

```python

# CODE_RUNNER: Add game scores to the log until a -1 sentinel value appears.

game_log = []
next_scores = [24, 31, 28, -1]
i = 0

while next_scores[i] != -1:
    game_log.append(next_scores[i])
    i += 1

print("Game log: " + str(game_log))
```

```python
%%pseudocode

// CODE_RUNNER: Add game scores to the log until a -1 sentinel value appears.

gameLog <-- []
nextScores <-- [24, 31, 28, -1]
i <-- 1

REPEAT UNTIL (nextScores[i] = -1) {
    APPEND(gameLog, nextScores[i])
    i <-- i + 1
}

DISPLAY(gameLog)
```

```python
%%js

// CODE_RUNNER: Add game scores to the log until a -1 sentinel value appears.

const gameLog = [];
const nextScores = [24, 31, 28, -1];
let i = 0;

while (nextScores[i] !== -1) {
  gameLog.push(nextScores[i]);
  i++;
}

console.log("Game log: " + gameLog);
```

### Algorithm

**CPT Requirement:** combine sequence, selection, and iteration. Average points per game, then rate it.

```python

# CODE_RUNNER: Average points per game, then assign a rating.

def average_points(games):
    total = 0
    for g in games:
        total += g
    return total / len(games)

game_points = [22, 34, 27, 30]
avg = average_points(game_points)

rating = "Needs Review"
if avg >= 30:
    rating = "Excellent"
elif avg >= 25:
    rating = "Strong"

print("Average Points: " + str(avg))
print("Rating: " + rating)
```

```python
%%pseudocode

// CODE_RUNNER: Average points per game, then assign a rating.

PROCEDURE averagePoints(games) {
    total <-- 0
    FOR EACH game IN games {
        total <-- total + game
    }
    RETURN total / LENGTH(games)
}

average <-- averagePoints([22, 34, 27, 30])
rating <-- "Needs Review"
IF (average >= 30) {
    rating <-- "Excellent"
} ELSE {
    IF (average >= 25) {
        rating <-- "Strong"
    }
}
DISPLAY("Average Points: " + average)
DISPLAY("Rating: " + rating)
```

```python
%%js

// CODE_RUNNER: Average points per game, then assign a rating.

function averagePoints(games) {
  let total = 0;
  for (const game of games) total += game;
  return total / games.length;
}

const average = averagePoints([22, 34, 27, 30]);
const rating = average >= 30 ? "Excellent" : average >= 25 ? "Strong" : "Needs Review";
console.log("Average Points: " + average);
console.log("Rating: " + rating);
```

### Search Algorithm

**Exam Topic:** linear search. Look up a player's points by name.

```python

# CODE_RUNNER: Search the roster for a player's points using linear search.

def find_points(names, points, target):
    for i in range(len(names)):
        if names[i] == target:
            return points[i]
    return -1

names = ["Jordan Lee", "Sam Diaz", "Taylor Reed", "Alex Kim"]
points = [1540, 980, 1120, 1375]

search_name = "Taylor Reed"
result = find_points(names, points, search_name)

if result != -1:
    print(search_name + "'s points: " + str(result))
else:
    print(search_name + " is not on the roster.")
```

```python
%%pseudocode

// CODE_RUNNER: Search the roster for a player's points using linear search.

PROCEDURE findPoints(names, points, target) {
    i <-- 1
    FOR EACH name IN names {
        IF (name = target) {
            RETURN points[i]
        }
        i <-- i + 1
    }
    RETURN -1
}

names <-- ["Jordan Lee", "Sam Diaz", "Taylor Reed", "Alex Kim"]
points <-- [1540, 980, 1120, 1375]
result <-- findPoints(names, points, "Taylor Reed")
DISPLAY(result)
```

```python
%%js

// CODE_RUNNER: Search the roster for a player's points using linear search.

function findPoints(names, points, target) {
  for (let i = 0; i < names.length; i++) {
    if (names[i] === target) return points[i];
  }
  return -1;
}

const names = ["Jordan Lee", "Sam Diaz", "Taylor Reed", "Alex Kim"];
const points = [1540, 980, 1120, 1375];
const searchName = "Taylor Reed";
const result = findPoints(names, points, searchName);
console.log(result !== -1 ? searchName + "'s points: " + result : searchName + " is not on the roster.");
```

### Boolean Logic

**Exam Concept:** combine conditions with AND, OR, and NOT. Decide whether a player makes the All-Star team.

```python

# CODE_RUNNER: Decide if a player is an All-Star using AND / OR / NOT.

points_per_game = 28
games_played = 15
coach_recommendation = False

all_star_selected = (points_per_game >= 25 and games_played > 10) or coach_recommendation
is_rookie = not (games_played > 5)

print("All-Star selected: " + str(all_star_selected))
print("Rookie: " + str(is_rookie))
```

```python
%%pseudocode

// CODE_RUNNER: Decide if a player is an All-Star using AND / OR / NOT.

pointsPerGame <-- 28
gamesPlayed <-- 15
coachRecommendation <-- FALSE
allStarSelected <-- (pointsPerGame >= 25 AND gamesPlayed > 10) OR coachRecommendation
isRookie <-- NOT (gamesPlayed > 5)
DISPLAY("All-Star selected: " + allStarSelected)
DISPLAY("Rookie: " + isRookie)
```

```python
%%js

// CODE_RUNNER: Decide if a player is an All-Star using AND / OR / NOT.

const pointsPerGame = 28;
const gamesPlayed = 15;
const coachRecommendation = false;
const allStarSelected = (pointsPerGame >= 25 && gamesPlayed > 10) || coachRecommendation;
const isRookie = !(gamesPlayed > 5);
console.log("All-Star selected: " + allStarSelected);
console.log("Rookie: " + isRookie);
```
