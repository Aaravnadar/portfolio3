### Homework Hack: Music Practice Panel

Four toggles, each with its own label and job. JavaScript reads `.checked` on every one, a live counter shows how many are on, and three of the toggles change real parts of the page: a tip that shows or hides, a night mode preview, and larger text. I built the toggles from a list with a loop, and the comments explain what I learned.

{% capture challenge3 %}
Homework Hack: build your own toggles and react to them with JavaScript. Press Run and flip each one.
{% endcapture %}

{% capture code3 %}
// Build the toggles from a list, then wire them up. Edit, then press Run.
const settings = [
  { id: 'tip', label: 'Show practice tip' },
  { id: 'night', label: 'Night mode' },
  { id: 'big', label: 'Large sheet text' },
  { id: 'alerts', label: 'Practice alerts' }
];

// Loop over the list so each toggle is built the same way with its own id and label
let toggleMarkup = '';
settings.forEach(function (s) {
  toggleMarkup += '<label class="ocs__toggle ocs__toggle--green">' +
    '<input id="mp-' + s.id + '" class="ocs__toggle-input" type="checkbox">' +
    '<span class="ocs__toggle-track" aria-hidden="true"></span>' +
    '<span class="ocs__toggle-label">' + s.label + '</span>' +
    '</label>';
});

outputElement.innerHTML = '<div id="mp-panel">' + toggleMarkup +
  '<p id="mp-tip-text" hidden>Tip: play it slow first, then speed up a little each pass.</p>' +
  '<div id="mp-sheet">Sheet music preview: this box follows the toggles above.</div>' +
  '<p id="mp-alert-line"></p>' +
  '<p id="mp-total" aria-live="polite"></p>' +
  '</div>';

// What I learned: .checked is a true/false value that the checkbox already stores,
// so my code only has to read it and never has to remember the state itself.
const tipBox = outputElement.querySelector('#mp-tip');
const nightBox = outputElement.querySelector('#mp-night');
const bigBox = outputElement.querySelector('#mp-big');
const alertsBox = outputElement.querySelector('#mp-alerts');

const tipText = outputElement.querySelector('#mp-tip-text');
const sheet = outputElement.querySelector('#mp-sheet');
const alertLine = outputElement.querySelector('#mp-alert-line');
const totalLine = outputElement.querySelector('#mp-total');

// A single refresh function keeps the whole panel in sync with the toggles
function refresh() {
  // Toggle 1 shows or hides the tip (hidden is the opposite of checked)
  tipText.hidden = !tipBox.checked;

  // Toggle 2 swaps the preview between a dark and a light theme
  sheet.style.padding = '12px';
  sheet.style.background = nightBox.checked ? '#111827' : '#fef9c3';
  sheet.style.color = nightBox.checked ? '#f9fafb' : '#1f2937';

  // Toggle 3 makes the preview text bigger or normal
  sheet.style.fontSize = bigBox.checked ? '22px' : '15px';

  // Toggle 4 only changes a status sentence
  alertLine.textContent = alertsBox.checked ? 'Practice alerts: ON' : 'Practice alerts: OFF';

  // Count the toggles that are on. For every toggle, IF it is checked THEN add one,
  // which is the same selection idea used in AP CSP.
  const all = outputElement.querySelectorAll('.ocs__toggle-input');
  let turnedOn = 0;
  all.forEach(function (box) {
    if (box.checked) {
      turnedOn = turnedOn + 1;
    }
  });
  totalLine.textContent = 'Toggles on right now: ' + turnedOn + ' out of ' + all.length;
}

// The change event fires whenever a toggle flips, which keeps the count live
[tipBox, nightBox, bigBox, alertsBox].forEach(function (box) {
  box.addEventListener('change', refresh);
});

// Call it once at the start so the page begins with the right text (0 out of 4)
refresh();

**How it meets each requirement**

- **3 or more toggles with their own label and purpose:** four toggles (practice tip, night mode, large sheet text, practice alerts).
- **JavaScript checks every on/off state:** `.checked` is read for each toggle inside `refresh()`.
- **A live count of toggles that are on:** the line `Toggles on right now: N out of 4` updates whenever the `change` event fires.
- **At least 2 comments that show understanding:** the comments explain `.checked`, why one `refresh()` function keeps everything in sync, and how the counting loop is a selection statement from AP CSP.
- **Above and beyond:** the toggles are generated from a list with a loop, and three of them drive real features (tip, night mode, large text) instead of only changing their own look.
