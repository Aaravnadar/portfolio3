## Popcorn Hack: Ocean Buoy Wave Scanner

Skip calm readings with `continue`, count and total the readings that need attention, and end the scan at the first dangerous reading with `break`. The calm cutoff is **4** (anything lower is skipped) and the danger cutoff is **9**.

````python
# CODE_RUNNER: Popcorn Hack - ocean buoy wave scanner
# Theme: A research team scans wave heights reported by different buoys.
# Goal: skip calm readings, count readings that need attention, total their scores, and stop at the first dangerous reading.

waves = [2, 7, 1, 5, 4, 10, 6]

# Readings below this number are skipped.
calm_cutoff = 4

# The scan stops at this number or higher.
danger_cutoff = 9

# Counters and totals begin at 0 because nothing has been read yet.
attention_count = 0
calm_count = 0
attention_total = 0
danger_found = False

# Go through each wave reading in order.
for wave in waves:
    # Calm reading: count it, print it, and jump to the next reading.
    if wave < calm_cutoff:
        calm_count += 1
        print("Wave " + str(wave) + ": calm, skipping")
        continue

    # Reaching this line means the reading needs attention.
    attention_count += 1
    attention_total += wave
    print("Wave " + str(wave) + ": needs attention")

    # Dangerous reading: record it and end the scan.
    if wave >= danger_cutoff:
        danger_found = True
        print("Dangerous wave found, ending scan")
        break

# Final report that sums up the scan.
print("Calm readings skipped: " + str(calm_count))
print("Readings needing attention: " + str(attention_count))
print("Attention total: " + str(attention_total))
print("Danger found: " + str(danger_found))
````

**Actual output:** waves `2` and `1` are skipped as calm; `7`, `5`, `4`, and `10` need attention; the scan ends at `10`, so `6` is never read. Final report: calm readings skipped `2`, readings needing attention `4`, attention total `26`, danger found `True`.

**Explanation:** `continue` only skips the rest of the current pass, which fits calm readings. `break` leaves the whole loop, which fits the first dangerous reading. The counters start at `0` and grow with `+=` inside the loop.

## Homework Hack: Themed Loop-Based Checker

**Theme: classroom noise monitoring.** Readings below `40` are skipped and readings of `90` or more are extreme and stop the scan. The program has eight values, two thresholds, a `for` loop, `continue`, `break`, and five tracked summary values.

````python
# CODE_RUNNER: Homework Hack - themed loop-based checker
# Theme: classroom noise monitoring (decibels)
# Goal: skip quiet readings, count loud readings, total their scores, and stop at the first extreme reading.

noise_levels = [35, 52, 28, 61, 44, 70, 95, 58]

# Readings below quiet_cutoff are skipped with continue
quiet_cutoff = 40

# Readings at or above extreme_cutoff end the loop with break
extreme_cutoff = 90

# Summary variables
looked_at = 0
loud_count = 0
quiet_count = 0
loud_total = 0
extreme_found = False

# Walk through the readings until an extreme one stops the scan
for level in noise_levels:
    # Every reading the loop touches gets counted
    looked_at += 1

    # Quiet readings are tallied and then skipped
    if level < quiet_cutoff:
        quiet_count += 1
        print(str(level) + " dB: quiet, no action")
        continue

    # Reaching this line means the reading is loud enough to act on
    loud_count += 1
    loud_total += level
    print(str(level) + " dB: loud, action needed")

    # End the scan early on an extreme reading
    if level >= extreme_cutoff:
        extreme_found = True
        print("Extreme reading found. Stop checking.")
        break

# Final report
print("Final Report")
print("Looked at: " + str(looked_at))
print("Quiet: " + str(quiet_count))
print("Loud: " + str(loud_count))
print("Loud total: " + str(loud_total))
print("Extreme found: " + str(extreme_found))
````

**Actual output:** `35` and `28` are skipped, `52`, `61`, `44`, `70`, and `95` need action, and the scan stops at `95` (so the last value, `58`, is never looked at).

Final Report: Looked at `7`, Quiet `2`, Loud `5`, Loud total `322`, Extreme found `True`.
