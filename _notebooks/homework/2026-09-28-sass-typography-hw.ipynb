### Homework Hack: How to Make Toast

- Pretend headings turn into real ones: `<h2>` for the page title, then `<h3>` for each section, one level at a time.
- Each `body-text` div becomes a proper `<p>`.
- The bold `very-important` span becomes a `<strong>` tag.
- Three lines that were numbered by hand collapse into **one** `<ol>` with three `<li>` items, so the browser handles the numbers.
- The whole page sits in an `ocs_card`, with the title using `ocs_section-title`. There is no `<style>` block, no inline style, and no custom class.
