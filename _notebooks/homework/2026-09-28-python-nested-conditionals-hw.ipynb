## Popcorn Hack 1: Commute Plan (Beginner)

**Trace:** `is_raining` is `True`, so the outer `if` enters the rainy path. Then `has_umbrella` is `False`, so the inner `else` runs.

**Prediction:** `Take the bus`  
**Outer branch:** `if is_raining` (True) **Inner branch:** `else` under `if has_umbrella` (False)

````python
# CODE_RUNNER: Popcorn Hack 1 - Test the commute plan

is_raining = True
has_umbrella = False

if is_raining:
    if has_umbrella:
        print("Walk with your umbrella")
    else:
        print("Take the bus")
else:
    print("Enjoy the walk")
````

**Actual output:** `Take the bus`

**Reaching the other two outcomes:** `is_raining = True, has_umbrella = True` prints `Walk with your umbrella`. `is_raining = False` (the umbrella value does not matter) prints `Enjoy the walk`. The next cell runs all three paths.

````python
# CODE_RUNNER: Popcorn Hack 1 - All three paths

tests = [(True, False), (True, True), (False, False)]

for is_raining, has_umbrella in tests:
    if is_raining:
        if has_umbrella:
            print("Walk with your umbrella")
        else:
            print("Take the bus")
    else:
        print("Enjoy the walk")
````

**Actual output:** `Take the bus`, `Walk with your umbrella`, `Enjoy the walk`.

## Popcorn Hack 2: Repair the Door Scanner (Intermediate)

**Why the original was not nested:** the second `if balance_ok:` was written at the *same* indentation as `if card_scanned:`, which makes it a separate, independent decision. The balance got checked even when no card had been scanned. Indentation is what ties one decision to another.

**Fix:** the balance check moves inside the scanned branch, and an outer `else` prints `Scan your card`.

````python
# CODE_RUNNER: Popcorn Hack 2 - Repair the door scanner

card_scanned = True
balance_ok = False

if card_scanned:
    print("Card read")
    if balance_ok:
        print("Door unlocked")
    else:
        print("Add funds")
else:
    print("Scan your card")
````

**Actual output:** `Card read`, then `Add funds`.

### Test all three messages

````python
# CODE_RUNNER: Popcorn Hack 2 - Test all three outcomes

tests = [(True, True), (True, False), (False, False)]

for card_scanned, balance_ok in tests:
    if card_scanned:
        print("Card read")
        if balance_ok:
            print("Door unlocked")
        else:
            print("Add funds")
    else:
        print("Scan your card")
    print("---")
````

**Actual output:** `Card read` / `Door unlocked`, then `Card read` / `Add funds`, then `Scan your card`. There are three possible results, and after a successful scan only one of the two balance messages can appear.

## Popcorn Hack 3: Bike Ride Plan (Advanced)

`bike_working` is checked **first** (outer). `is_dark_out` is checked inside each branch (inner), so the second question is only asked once the bike decision has chosen a path.

````python
# CODE_RUNNER: Popcorn Hack 3 - Test the bike ride plan

tests = [(True, True), (True, False), (False, True), (False, False)]

for bike_working, is_dark_out in tests:
    if bike_working:
        if is_dark_out:
            print("Ride your bike and turn on your lights")
        else:
            print("Ride your bike, no lights needed")
    else:
        if is_dark_out:
            print("Take the bus and wait at a lit stop")
        else:
            print("Take the bus, no extra steps")
````

**Actual output:** four different suggestions, one for each combination (bike + dark, bike + daylight, bus + dark, bus + daylight).

## Homework Hack: Student Club Day Planner

Outer decision: is the member logged in? Inside that: is the project finished? Inside *that*: is there a meeting today, and (when there is) does the member have a ride? The starter `member` dictionary is the first test case, followed by cases that reach every other outcome.

````python
# CODE_RUNNER: Homework Hack - Build a student club day planner

member = {
    "logged_in": True,
    "project_done": False,
    "has_ride": True,
    "meeting_today": True
}

# Test cases: the starter member first, then one case for every other outcome
test_members = [
    member,
    {"logged_in": False, "project_done": False, "has_ride": False, "meeting_today": False},
    {"logged_in": True, "project_done": False, "has_ride": True, "meeting_today": False},
    {"logged_in": True, "project_done": True, "has_ride": True, "meeting_today": True},
    {"logged_in": True, "project_done": True, "has_ride": False, "meeting_today": True},
    {"logged_in": True, "project_done": True, "has_ride": True, "meeting_today": False},
]

for m in test_members:
    # Outer decision: nothing else matters until the member is logged in
    if m["logged_in"]:
        # Inner decision 1: the project comes before any meeting plans
        if m["project_done"]:
            # Inner decision 2: a ride only matters when there is a meeting
            if m["meeting_today"]:
                # Inner decision 3: the travel advice depends on the ride
                if m["has_ride"]:
                    print("Project done. Ride to the meeting.")
                else:
                    print("Project done. Ask for a ride or take the bus.")
            else:
                print("Project done and no meeting. Free evening.")
        else:
            if m["meeting_today"]:
                print("Finish the project before the meeting.")
            else:
                print("Finish the project tonight.")
    else:
        print("Log in first to see your planner.")
````

**Actual output (in order):** `Finish the project before the meeting.` / `Log in first to see your planner.` / `Finish the project tonight.` / `Project done. Ride to the meeting.` / `Project done. Ask for a ride or take the bus.` / `Project done and no meeting. Free evening.`

**Why the inner checks depend on the outer ones:** the planner can only discuss projects, meetings, or rides once the login check passes, so those checks sit *inside* the logged-in branch. A ride only matters when a meeting exists, and a meeting plan only matters once the project is finished, so each deeper question depends on the answer above it. This is different from one long `and` condition because each path ends with its own message.

**Six outcomes tested** (one not logged in, two with the project unfinished, three with the project finished), which covers every meaningful path.
