## Popcorn Hack 1: Read the API, Then Use It

**What I changed:** the random number is now a locker from `100` to `450`, and the square area is `400` square feet.

**Prediction:** `random.randint(100, 450)` returns any whole number from 100 through 450 (both included), so the locker changes on every run. `math.sqrt(400)` is `20.0`, so the garden side prints `20.0 feet`. The average of `[88, 92, 79, 95, 84]` is `87.6`, and `math.floor(87.6)` rounds down to `87`.

````python
# CODE_RUNNER: Popcorn Hack 1 - Read the API, then use it (locker 100-450, bigger garden)

import math
import random

# Read the API before calling a procedure
print("random.randint:", random.randint.__doc__)
print("math.floor:", math.floor.__doc__)
print()

quiz_scores = [88, 92, 79, 95, 84]

# random.choice picks one item out of a list
print("Random quiz to review:", random.choice(quiz_scores), "points")

# CHANGED: randint(a, b) now returns a whole number from 100 to 450, both included
locker = random.randint(100, 450)
print("Random locker number:", locker)

# CHANGED: a 400 square foot garden bed instead of a smaller one
side = math.sqrt(400)
print("Square garden side:", side, "feet")

# math.floor rounds down to the previous whole number
average = sum(quiz_scores) / len(quiz_scores)
rounded_down = math.floor(average)
print("Quiz average:", average, "-> rounded down:", rounded_down)
````

**Actual output:** the docstrings for `randint` and `floor`, a random quiz score from the list, a locker number between 100 and 450, `Square garden side: 20.0 feet`, and `Quiz average: 87.6 -> rounded down: 87`.

The random lines change each run but always stay inside the range the API promises. That is why reading the docs matters: I only needed to know what each procedure accepts and returns, not how it works inside.

## Popcorn Hack 2: Write a Flask Route

A route is a procedure with a web address on top. I built a small cafeteria app and added my own `/hours` route. The routes are defined above `app.test_client()`.

````python
# CODE_RUNNER: Popcorn Hack 2 - Flask routes for a cafeteria app with a /hours route

from flask import Flask, jsonify

app = Flask(__name__)

# A route is a procedure with a web address attached to it
@app.route("/menu")
def menu():
    return jsonify({"lunch": "Veggie Wrap", "side": "Fruit Cup", "price": 5.5})

@app.route("/specials")
def specials():
    return jsonify({"specials": ["Tomato Soup", "Grilled Cheese", "Mango Smoothie", "Rice Bowl"]})

# MY EXTRA ROUTE: /hours
@app.route("/hours")
def hours():
    return jsonify({"opens": "11:00 AM", "closes": "2:00 PM"})

# test_client visits the addresses without needing a browser
client = app.test_client()

print("/menu says:", client.get("/menu").get_json())
print("/specials says:", client.get("/specials").get_json())
print("/hours says:", client.get("/hours").get_json())
````

**Actual output:**

```
/menu says: {'lunch': 'Veggie Wrap', 'price': 5.5, 'side': 'Fruit Cup'}
/specials says: {'specials': ['Tomato Soup', 'Grilled Cheese', 'Mango Smoothie', 'Rice Bowl']}
/hours says: {'closes': '2:00 PM', 'opens': '11:00 AM'}
```

**My extra route:** `/hours`. (The keys print in alphabetical order because `jsonify` sorts them.)

## Homework Hack: Reading Log Report

A reading log replaces the sports sample, with my own sessions and my own next-session date. It uses three standard library modules: `json`, `statistics`, and `datetime`. The `summarize_log` procedure has a docstring that explains its API, so someone could use it without reading the code.

````python
# CODE_RUNNER: Homework Hack - Reading Log Report

import json
import statistics
from datetime import date

sessions = [
    {"date": "2026-10-05", "book": "Dune", "pages": 32, "minutes": 45},
    {"date": "2026-10-09", "book": "Project Hail Mary", "pages": 41, "minutes": 55},
    {"date": "2026-10-14", "book": "Dune", "pages": 28, "minutes": 40},
    {"date": "2026-10-19", "book": "The Hobbit", "pages": 50, "minutes": 70},
    {"date": "2026-10-24", "book": "Project Hail Mary", "pages": 36, "minutes": 50},
]

def summarize_log(log):
    """
    Summarize a list of reading sessions.

    Parameters:
        log: list of dictionaries, each with "date", "book", "pages", and "minutes"
    Returns:
        dictionary with total_pages, average_pages, longest_session,
        and pages_by_book
    """
    pages = [entry["pages"] for entry in log]

    # Add up the pages read from each book
    pages_by_book = {}
    for entry in log:
        title = entry["book"]
        pages_by_book[title] = pages_by_book.get(title, 0) + entry["pages"]

    return {
        "total_pages": sum(pages),
        "average_pages": round(statistics.mean(pages), 1),
        "longest_session": max(pages),
        "pages_by_book": pages_by_book,
    }

report = summarize_log(sessions)

# Subtracting two dates gives a timedelta, and .days is the whole number of days
next_session = date(2026, 11, 20)
report["days_until_next_session"] = (next_session - date.today()).days

print(json.dumps(report, indent=2))

# requirements.txt only lists EXTERNAL packages. json, datetime, and statistics
# ship with Python, so they never appear in it.
requirements_txt = """pandas
matplotlib
"""
print()
print("requirements.txt:")
print(requirements_txt)
````

**Actual output** (the `days_until_next_session` value depends on the day you run it; it is 21 on October 30, 2026):

```json

  "average_pages": 37.4,

  "pages_by_book": {
    "Dune": 60,
    "Project Hail Mary": 77,
    "The Hobbit": 50
  },
  "days_until_next_session": 21
}
```


`pandas` would hold the sessions as a table and `matplotlib` would draw a pages-per-day chart. `json`, `datetime`, and `statistics` are **not** listed because they are part of Python's standard library, so they come installed with Python and never need `pip install`.
