| --- | --- | --- |

| Adoption | `ocs__btn alert-green fill pill` | Green and solid for a happy, encouraging action |
| Events | `ocs__btn accent outline pill` | Outline only, since it matters but is not urgent |

## Homework Hack #2: Primary vs. Secondary Action Priority

The adoption panel is one row inside `ocs__links`. Visual weight tells the visitor what to do first:

| Action | Priority | Classes |
| --- | --- | --- |
| Start Adoption Form | Primary | `ocs__btn accent fill` |
| See Available Pets | Secondary | `ocs__btn outline` |
| Report Animal in Danger | Alert | `ocs__btn alert-red outline` |

"Start Adoption Form" is a `<button>` because it triggers something on this page. The other two are `<a>` tags because they send the visitor somewhere else.

{% capture challenge5 %}
Homework Hack 2: one filled primary action, one outline secondary action, and one red outline alert action in ocs__links. Press Run.
{% endcapture %}

{% capture code5 %}
// Edit the HTML inside the quotes, then press Run.

  '<div class="ocs__links">',

  '  <a class="ocs__btn alert-red outline" href="https://example.org/danger">Report Animal in Danger</a>',
  '</div>'
].join('');
