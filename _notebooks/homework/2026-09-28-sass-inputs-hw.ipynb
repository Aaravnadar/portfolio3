### Homework Hack: Refactor the Volunteer Form

- The 500px pink-to-purple field turns into `ocs__input large gradient`.
- The `tinybox` field turns into `ocs__input small`.
- The email field loses its inline padding and uses `ocs__input medium`. It keeps `type="email"`, so the browser still checks the format.
- Every field has a placeholder that explains its job, plus an `aria-label` so screen readers can announce it.
