## Homework Hack: Parade Info Page

The Poway Rotary Parade info page includes:

* A table built from `ocs__table-wrap` + `ocs__table`, with one header row and two body rows
* A `cols-3` standard grid that has a `--header` cell, an `--accent` cell, and a `--muted` cell
* An `ocs__section-title` plus a single-line `ocs__description`
* A closing `ocs__callout` line
* All of it inside one `ocs__card`, which sits in one `ocs__container`
