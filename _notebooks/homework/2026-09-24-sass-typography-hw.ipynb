## Homework Hack

The "How to Pet a Cat" guide is rebuilt inside an OCS card using semantic tags:

* The three steps: a single `<ol>` holding three `<li>`
* Key text: `<strong>`
* Regular text: `<p class="ocs__text">`
* Section headings: `<h3>`
* Main title: `<h2 class="ocs__section-title">`

{% capture challenge2 %}
Homework Hack: the cat guide using semantic tags, strong text, an ordered list, and OCS classes. Click Run.
{% endcapture %}

{% capture code2 %}
// Change the HTML between the quotes, then click Run.
outputElement.innerHTML = [
  '<div class="ocs__card">',
  '  <h2 class="ocs__section-title">How to Pet a Cat</h2>',
  '  <h3>Ever wanted to…</h3>',
  '  <p class="ocs__text">pet a kitty? Lucky you, here is a simple step by step guide to petting a cat!</p>',
  '  <p class="ocs__text"><strong>Steps to pet a cat:</strong></p>',
  '  <ol>',
  '    <li>hold out your hand</li>',
  '    <li>gently touch the cat</li>',
  '    <li>slowly stroke back and forth</li>',
  '  </ol>',
  '  <h3>Nice job! You pet a cat :D</h3>',
  '</div>'
].join('');
