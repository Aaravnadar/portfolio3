## Popcorn Hack #1: Add the Base Button Class

The empty `class=""` is replaced with `ocs__btn` so the link gets OCS button styling.

```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 1 Base -->

<a class="ocs__btn" href="https://www.mygoodbrain.org/resources">Mental Health Resource Directory</a>
```

**Why:** `ocs__btn` is the base button role. It turns a plain `<a>` into a styled button with zero custom CSS.

## Popcorn Hack #2: Modifiers & Button Groups

`pill` and `fill` are added to the check-in button, and both actions are wrapped in an `ocs__links` container.

```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 2 Modifiers -->

<div class="ocs__links">
  <button class="ocs__btn pill accent fill">Complete Daily Check-In</button>
  <a class="ocs__btn outline" href="">Save Draft Journal</a>
</div>
```

**Why:** `pill` rounds the corners, `fill` makes it solid, and `ocs__links` handles spacing so no extra CSS is needed.

## Popcorn Hack #3: Status Tones & Wide Layout Bar
