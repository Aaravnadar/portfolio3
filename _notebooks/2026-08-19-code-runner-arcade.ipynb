### Procedure

```python
%%js

// CODE_RUNNER: Write a procedure that returns a rank title based on score.

function getRankTitle(score) {
  if (score >= 15000) {
    return "Legend";
  } else if (score >= 10000) {
    return "Pro";
  } else {
    return "Rookie";
  }
}

const myScore = 12300;
console.log("Rank: " + getRankTitle(myScore));
```

### Sequence

### Selection

**CPT Requirement:** the algorithm must include IF / ELSE decisions. Assign a medal based on a player's final score.

### Iteration

**CPT Requirement:** the algorithm must include a loop. Keep adding scores to a leaderboard until a sentinel value (-1) appears -- a REPEAT UNTIL pattern.

```python
%%js

// CODE_RUNNER: Add scores to the leaderboard until a -1 sentinel value appears.

const leaderboard = [];
const nextEntries = [8700, 11200, 9800, -1];
let i = 0;

while (nextEntries[i] !== -1) {
  leaderboard.push(nextEntries[i]);
  i++;
}

console.log("Leaderboard: " + leaderboard);
```

```python
%%python

# CODE_RUNNER: Add scores to the leaderboard until a -1 sentinel value appears.

leaderboard = []
next_entries = [8700, 11200, 9800, -1]
i = 0

while next_entries[i] != -1:
    leaderboard.append(next_entries[i])
    i += 1

print("Leaderboard: " + str(leaderboard))
```

### Algorithm

**CPT Requirement:** combine sequence, selection, and iteration into one algorithm. Average a set of round scores, then assign a medal from the average.

```python
%%js

// CODE_RUNNER: Average a set of round scores, then assign a medal.

function averageScore(scores) {
  let total = 0;
  for (const s of scores) {
    total += s;
  }
  return total / scores.length;
}

const roundScores = [9200, 14500, 11800, 13300];
const avg = averageScore(roundScores);

let medal = "None";
if (avg >= 13000) {
  medal = "Gold";
} else if (avg >= 10000) {
  medal = "Silver";
}

console.log("Average Score: " + avg);
console.log("Medal: " + medal);
```

```python
%%python

# CODE_RUNNER: Average a set of round scores, then assign a medal.

def average_score(scores):
    total = 0
    for s in scores:
        total += s
    return total / len(scores)

round_scores = [9200, 14500, 11800, 13300]
avg = average_score(round_scores)

medal = "None"
if avg >= 13000:
    medal = "Gold"
elif avg >= 10000:
    medal = "Silver"

print("Average Score: " + str(avg))
print("Medal: " + medal)
```

### Search Algorithm

**Exam Topic:** linear search through a list. Look up a player's score by name in the roster.

```python
%%js

// CODE_RUNNER: Search the roster for a player's score using linear search.

function findScore(names, scores, target) {
  for (let i = 0; i < names.length; i++) {
    if (names[i] === target) {
      return scores[i];
    }
  }
  return -1;
}

const names = ["Nova", "Blitz", "Rook", "Echo"];
const scores = [15400, 9800, 11200, 13750];

const searchName = "Rook";
const result = findScore(names, scores, searchName);

if (result !== -1) {
  console.log(searchName + "'s score: " + result);
} else {
  console.log(searchName + " not found on the roster.");
}
```

```python
%%python

# CODE_RUNNER: Search the roster for a player's score using linear search.

def find_score(names, scores, target):
    for i in range(len(names)):
        if names[i] == target:
            return scores[i]
    return -1

names = ["Nova", "Blitz", "Rook", "Echo"]
scores = [15400, 9800, 11200, 13750]

search_name = "Rook"
result = find_score(names, scores, search_name)

if result != -1:
    print(search_name + "'s score: " + str(result))
else:
    print(search_name + " not found on the roster.")
```

### Boolean Logic

**Exam Concept:** combine conditions with AND, OR, and NOT. Decide whether a player unlocks the bonus round.

```python
%%js

// CODE_RUNNER: Decide if a player unlocks the bonus round using AND / OR / NOT.

const score = 13750;
const livesRemaining = 2;
const hasSecretCoin = false;

const unlocksBonusRound = (score >= 12000 && livesRemaining > 0) || hasSecretCoin;
const isEliminated = !(livesRemaining > 0);

console.log("Unlocks bonus round: " + unlocksBonusRound);
console.log("Eliminated: " + isEliminated);
```

```python
%%python

# CODE_RUNNER: Decide if a player unlocks the bonus round using AND / OR / NOT.

score = 13750
lives_remaining = 2
has_secret_coin = False

unlocks_bonus_round = (score >= 12000 and lives_remaining > 0) or has_secret_coin
is_eliminated = not (lives_remaining > 0)

print("Unlocks bonus round: " + str(unlocks_bonus_round))
print("Eliminated: " + str(is_eliminated))
```
